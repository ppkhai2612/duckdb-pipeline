```python
import duckdb
```

```python
con = duckdb.connect()
```

```python
# ensure httpfs extension is loaded!
con.load_extension("httpfs")
```

```python
process_date='2026-10-22'
gold_url=f"s3://gharchive-gold/gharchive/events/{process_date}/*.parquet"
```

```python
con.execute(f"SET s3_access_key_id = 'test'")
con.execute(f"SET s3_secret_access_key = 'test'")
con.execute(f"SET s3_region = 'us-east-1'")
con.execute(f"SET s3_endpoint = 'localhost:4566'")
con.execute("SET s3_url_style = 'path'")
con.execute("SET s3_use_ssl = true")
```

```python
con.sql(f"SELECT * FROM '{gold_url}'").to_table("gharchive_gold")
gharchive_gold = con.table("gharchive_gold")
```

```python
gharchive_gold.aggregate("event_type, repo_name, sum(event_count) as event_count")\
    .filter("event_type = 'WatchEvent'")\
    .order("event_count desc")\
    .limit(10).df()
```
